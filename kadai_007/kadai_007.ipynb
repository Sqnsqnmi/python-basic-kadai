```python
# リスト：順序（0から始まる番号）で文字列を並べる
array = [
    "月曜日は晴れです",
    "火曜日は雨です",
    "水曜日は晴れです",
    "木曜日は晴れです",
    "金曜日は曇りです",
    "土曜日は曇りのち雨です",
    "日曜日は雷雨です",
]

# ディクショナリ：キーと値の組で持つ
dictionary = {
    "mon": "晴れ",
    "tue": "雨",
    "wed": "晴れ",
    "thu": "晴れ",
    "fri": "曇り",
    "sat": "曇りのち雨",
    "sun": "雷雨",
}

# 水曜日の天気予報を取り出して出力する
print(array[2])  # リストは番号で取り出す（水曜日は2番目）
print(dictionary["wed"])  # ディクショナリはキーで取り出す
```
