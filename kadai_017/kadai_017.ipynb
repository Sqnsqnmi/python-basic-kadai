```python
# グローバル変数：大人とみなす年齢（クラスや関数の中からも読める）
ADULT_AGE = 20

# 名前と年齢の属性を持つHumanクラス
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 年齢がADULT_AGE以上なら大人、そうでなければ大人でないと出力する
    def check_adult(self):
        if self.age >= ADULT_AGE:
            print(f"{self.name}さん（{self.age}歳）は大人です")
        else:
            print(f"{self.name}さん（{self.age}歳）は大人ではありません")

# 名前と年齢のディクショナリから、Humanインスタンスのリストを作って返す関数
def create_humans(people):
    humans = []  # ローカル変数：この関数の中でだけ使う
    for name, age in people.items():
        humans.append(Human(name, age))
    return humans

# 名前をキー、年齢を値にしたディクショナリ
people = {"侍太郎": 36, "侍花子": 19, "侍一郎": 20}

# Humanインスタンスを複数生成してリストに追加する
human_list = create_humans(people)

# リストの要素数分だけcheck_adultメソッドを呼び出す
for human in human_list:
    human.check_adult()
```
