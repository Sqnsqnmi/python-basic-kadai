```python
# 名前と年齢の属性を持つHumanクラス
class Human:
    # コンストラクタ：インスタンス化するときに属性に値を代入する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を出力するメソッド
    def printinfo(self):
        print(f"名前：{self.name}")
        print(f"年齢：{self.age}歳")

# Humanクラスのインスタンスを変数に代入する
human = Human("侍太郎", 36)

# メソッドを使って名前と年齢を出力する
human.printinfo()
```
